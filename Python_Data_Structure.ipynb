Python Data Structure

Q1. What are data structures, and why are they important?

Ans: Data structures are ways to organize and store data (like lists, stacks, queues, trees).
They are important because they make data handling efficient, enabling faster access, modification, and problem-solving.

Q2. Explain the difference between mutable and immutable data types with examples.

Ans: Mutable types can be changed after creation (e.g., list, dict, set).
Immutable types cannot be modified once created (e.g., tuple, str, int).

Q3. What are the main differences between lists and tuples in Python?

Ans: Lists are mutable, slower, and written with [], while tuples are immutable, faster, and written with ().

Q4. Describe how dictionaries store data.

Ans: Dictionaries store data as key–value pairs internally using a hash table, which allows fast lookups.

Q5. Why might you use a set instead of a list in Python?

   Ans: Sets automatically remove duplicate elements and allow faster membership checks compared to lists.

Q6. What is a string in Python, and how is it different from a list?

Ans: A string is an immutable sequence of characters, while a list is mutable and can hold elements of different data types.



Q7. How do tuples ensure data integrity in Python?

Ans: Tuples are immutable, so their values cannot be changed, preventing accidental modification and ensuring consistency.

Q8. What is a hash table, and how does it relate to dictionaries in Python?

Ans: A hash table stores data using a hash function to map keys to values. Python dictionaries are built on hash tables.

Q9. Can lists contain different data types in Python?

Ans: Yes, Python lists can hold mixed data types like integers, strings, and objects.

Q10. Explain why strings are immutable in Python.

Ans: Strings are immutable for memory efficiency, security, and to make them usable as dictionary keys.

Q11. What advantages do dictionaries offer over lists for certain tasks?

Ans: Dictionaries provide key-based access with average O(1) lookup, while lists require O(n) search by index.

Q12. Describe a scenario where using a tuple would be preferable over a list.

Ans: Tuples are better when storing fixed data like coordinates, dates, or when using them as dictionary keys.

Q13. How do sets handle duplicate values in Python?

Ans: Sets automatically remove duplicates and keep only unique elements.

Q14. How does the “in” keyword work differently for lists and dictionaries?

Ans: For lists, it checks if a value exists. For dictionaries, it checks if a key exists.

Q15. Can you modify the elements of a tuple? Explain why or why not.

Ans: No, tuples are immutable, so once created, their elements cannot be modified.

Q16. What is a nested dictionary, and give an example of its use case.

Ans: A nested dictionary is a dictionary inside another dictionary. Example: storing student details with subjects and marks.

Q17. Describe the time complexity of accessing elements in a dictionary.

Ans: Dictionary element access has average O(1) time complexity, but in the worst case it can degrade to O(n).

Q18. In what situations are lists preferred over dictionaries?

Ans: Lists are preferred when order matters, when storing sequential data, or when keys are not required.

Q19. Why are dictionaries considered unordered, and how does that affect data retrieval?

Ans: Dictionaries use hashing, so order wasn’t guaranteed before Python 3.7. This means retrieval is based on keys, not position.

Q20. Explain the difference between a list and a dictionary in terms of data retrieval.

Ans: Lists use index-based retrieval, while dictionaries use key-based retrieval for faster access.
